```python
import sys
import os

# Add the current directory to the Python path
sys.path.insert(0, os.getcwd())
sys.path.insert(0, os.path.join(os.getcwd(), 'backend'))
import torch

from strokes.stroke2img import strokes_to_image, save_as_bmp
from strokes.scale_strokes import rescale_strokes
from strokes.stroke_preprocessor import StrokePreprocessor

import numpy as np
from inference_single import Inference

```

```python
root = 'data'
file = 'test2'
STROKE_LENGTH = 25
```

```python
# get the strokes
parser = StrokePreprocessor()
label, strokes_array = parser.parse_inkml_strokes(f'{root}/{file}.inkml')
# Convert numpy array back to list of strokes for compatibility
strokes_list = []
unique_stroke_ids = sorted(set(strokes_array[:, 2]))

for stroke_id in unique_stroke_ids:
    mask = strokes_array[:, 2] == stroke_id
    stroke_points = strokes_array[mask][:, :2]  # Get x, y coordinates
    strokes_list.append(stroke_points.tolist())
strokes_list
```

```python
# rescale based on the stroke lengths
strokes_norm, size = rescale_strokes(strokes_list, STROKE_LENGTH)
print('Image size=', size)

# convert to image
img = strokes_to_image(strokes_norm, image_size=(int(size[0])+4, int(size[1])+4), line_thickness=2, padding=2)
save_as_bmp(img, 'test.bmp')
```

```python
inf = Inference(confPath='/Users/martijnslob/github/SAN/config.yaml')
latex = inf.convert2latex(img)
print("Predicted LaTeX:", latex)
```
